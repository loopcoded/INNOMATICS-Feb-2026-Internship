Feature	Stemming	Lemmatization
Method	Rule-based	Vocabulary + context
Output	Root (may be invalid)	Valid word
Example	"running" → "run"	"running" → "run"
Accuracy	Lower	Higher

Task 2: Advanced Preprocessing Function & Task 3: Stress Testing

```python
import re

def preprocess_text(text):
    try:
        if not isinstance(text, str) or text.strip() == "":
            return []

        # Remove URLs & emails
        text = re.sub(r"http\S+|www\S+|https\S+|\S+@\S+", "", text)

        # Remove numbers
        text = re.sub(r"\d+", "", text)

        # Lowercase
        text = text.lower()

        # Handle repeated characters (soooo → soo)
        text = re.sub(r'(.)\1{2,}', r'\1\1', text)

        # Remove special characters & emojis
        text = re.sub(r'[^a-z\s]', '', text)

        # Remove extra spaces
        text = re.sub(r'\s+', ' ', text).strip()

        tokens = text.split()

        # Remove short tokens except 'no', 'not'
        tokens = [word for word in tokens if len(word) > 2 or word in ['no', 'not']]

        return tokens

    except Exception as e:
        print("Error:", e)
        return []

test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for sentence in test_sentences:
    tokens = preprocess_text(sentence)
    clean_sentence = " ".join(tokens)

    results.append({
        "Original": sentence,
        "Tokens": tokens,
        "Cleaned": clean_sentence
    })

for r in results:
    print("\nOriginal:", r["Original"])
    print("Tokens:", r["Tokens"])
    print("Cleaned:", r["Cleaned"])
```

Task 4: Token Analytics

```python
import numpy as np

analytics = []

for r in results:
    tokens = r["Tokens"]

    total = len(tokens)
    unique = len(set(tokens))
    avg_len = np.mean([len(t) for t in tokens]) if tokens else 0

    analytics.append({
        "sentence": r["Original"],
        "total_tokens": total,
        "unique_tokens": unique,
        "avg_token_length": round(avg_len, 2)
    })

for a in analytics:
    print(a)
```

Task 5: Frequency Analysis

```python
from collections import Counter

all_tokens = []

for r in results:
    all_tokens.extend(r["Tokens"])

counter = Counter(all_tokens)

print("Top 10 Frequent Words:")
print(counter.most_common(10))

print("\nTop 5 Least Frequent Words:")
print(counter.most_common()[:-6:-1])
```

Task 6: Full Pipeline

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        clean_sentence = " ".join(tokens)

        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

pipeline_output = full_pipeline(test_sentences)

print("\nAll Tokens:\n", pipeline_output["tokens"])
print("\nClean Sentences:\n", pipeline_output["clean_sentences"])
```
